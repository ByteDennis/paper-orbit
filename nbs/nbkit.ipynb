# nbkit visual component gallery

A standalone HTML/CSS toolkit for richer VS Code notebooks. Run all cells to see the components in both light and dark themes.

```python
from importlib import reload
import nbkit

reload(nbkit)
from nbkit import (
    badge, callout, card, code, details, divider, grid, hero, kv,
    metric, metrics, panel, progress, quote, show, steps, table, tree,
)
```

## Hero and status language

```python
hero(
    "Notebook UI, without frontend scaffolding",
    "Composable, theme-aware HTML components built for VS Code and Jupyter output cells.",
)
grid(
    card("Design goal", "Dense technical information that still scans quickly.", accent=True, display_output=False),
    card("Runtime model", "Pure HTML and CSS; no JavaScript and no external assets.", display_output=False),
    columns=2,
)
```

```python
grid(
    badge("healthy", "success", display_output=False),
    badge("pending", "warning", display_output=False),
    badge("failed", "danger", display_output=False),
    badge("informational", "info", display_output=False),
    badge("experimental", "accent", display_output=False),
    columns=5,
)
```

## Callouts

```python
callout("Use callouts for context that changes how a reader interprets the next cell.", title="Note", tone="info")
callout("The validation suite completed without failures.", title="Ready", tone="success")
callout("This action replaces existing remote objects.", title="Review first", tone="warning")
callout("The source schema is incompatible with the manifest.", title="Blocked", tone="danger")
```

## Tables

```python
rows = [
    {"stage": "row_check", "state": "DONE", "rows": 128420, "seconds": 8.4},
    {"stage": "col_check", "state": "PARTIAL", "rows": 128420, "seconds": 22.1},
    {"stage": "report", "state": "FAILED", "rows": None, "seconds": 0.7},
]
table(rows, caption="Pipeline execution · automatic semantic colouring")
```

## Collapse / details

```python
details(
    "Open for the raw configuration",
    show(
        {
            "source": "oracle",
            "target": "main.finance.orders",
            "window": {"from": "2026-01-01", "to": "2026-01-31"},
        },
        "pair.json",
        display_output=False,
    ),
    tone="accent",
)
```

```python
details(
    "Expanded by default: validation evidence",
    [
        callout("All required fields are present.", title="Schema", tone="success", display_output=False),
        table(rows, caption="Evidence table inside details", display_output=False),
        code("SELECT COUNT(*) FROM main.finance.orders", "sql", "probe.sql", display_output=False),
    ],
    open=True,
    tone="success",
)
```

```python
details(
    "Nested details",
    details(
        "Inner diagnostic payload",
        show({"error_code": "TABLE_NOT_FOUND", "retryable": False}, display_output=False),
        tone="danger",
        display_output=False,
    ),
)
```

## Metrics and progress

```python
metrics(
    {
        "Rows compared": "1.28M",
        "Matched": "99.94%",
        "Findings": 17,
        "Runtime": "31.2s",
    },
    columns=4,
)
progress(86, label="Units reconciled", tone="success")
progress(34, label="Report publication", tone="warning")
```

## Key/value records and ordered steps

```python
grid(
    panel(
        kv(
            {
                "run_id": "live-20261005-a8f2c1",
                "profile": "corp",
                "workspace": "/Users/me/dtrack",
                "state": "RUNNING",
            },
            display_output=False,
        ),
        title="Run identity",
        display_output=False,
    ),
    panel(
        steps(
            [
                ("Discover", "Resolve profiles and source configuration."),
                ("Validate", "Run preflight without changing state."),
                ("Execute", "Publish units and reconcile results."),
                ("Report", "Write immutable artifacts."),
            ],
            display_output=False,
        ),
        title="Lifecycle",
        display_output=False,
    ),
    columns=2,
)
```

## Code, JSON, quote and tree

```python
code(
    '''def reconcile(left, right):
    return {"difference": left - right, "match": left == right}
''',
    "python",
    "reconcile.py",
)
show({"status": "DONE", "findings": ["row_count_diff", "left_only"]}, "summary.json")
quote("Good notebook UI makes evidence easier to inspect, not merely prettier.", "nbkit design principle")
```

## Composition recipe

```python
divider("one output, several components")
panel(
    [
        callout("The components below are composed before one display call.", tone="info", display_output=False),
        progress(72, label="Migration readiness", tone="accent", display_output=False),
        details(
            "Remaining checks",
            steps(["Oracle connectivity", "S3 write probe", "Databricks job smoke test"], display_output=False),
            display_output=False,
        ),
    ],
    title="Migration dashboard",
    subtitle="A reusable pattern for operational notebooks",
)
```
